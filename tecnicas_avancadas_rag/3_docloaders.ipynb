## Carregamento de documentos

### Carregamento de PDF

```python
documentos = PyPDFLoader(
    "./documentos/GTB_gold_Nov23.pdf"
).load()
```

```python
documentos
```

### Carregamento de página WEB

```python
url = "https://forbes.com.br/forbes-tech/2025/08/o-que-explica-o-fracasso-do-chatgpt-5-e-como-a-openai-vai-reagir/"

noticia = WebBaseLoader(web_path=url).load()
```

```python
noticia
```

### Leitura de diretório completo

```python
pdfs = DirectoryLoader("./documentos", glob="*.pdf").load()
```

```python
len(pdfs)
```

```python
pdfs[2]
```

### Merge dos documentos

```python
all_loaders = MergedDataLoader(
    loaders=[WebBaseLoader(web_path=url), TextLoader("./documentos/GTB_gold_Nov23.txt", encoding="utf-8")]
)

all_docs = all_loaders.load()
```

```python
len(all_docs)
```

```python
all_docs[0]
```

### Recursive URL Loader

```python
loader = RecursiveUrlLoader(
    "https://python.langchain.com/api_reference/"
)

lc_docs = loader.load()
```
